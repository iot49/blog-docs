## Beam Bending

Supported at ends, distributed load P [N/m]

[Beam Bending formulas](https://engineering.purdue.edu/~ce474/Docs/DA6-BeamFormulas.pdf)

Area Moment of Inertia:

$$I_a = \frac{tH^3}{12}$$

Maximum deflection:

$$\delta_{max} = \frac{5 P L^4}{384 E I_a}$$


```python
E = 10 * 1e9  # Young's modulus (e.g. 10 GPa)
L = 1.8  # Length of beam, [m]
H = 0.04  # Height of beam, [m], direction of load
t = 0.04  # thickness, [m], perpendicular to load
H = 150; t = 12e-3
load = 40  # load in [kg], distributed evenly over beam
P = 9.8*load/L  # Distributed load, [N/m]

Ia = t * H**3 / 12  # area Moment of inertia
delta_max = (5 * P * L**4) / (384 * E * Ia)

print(f"{L}m beam loaded with {load}kg (evenly distributed): deflection at center is {delta_max*1e3:.1f}mm")
```

```python
def delta_max(L: float, H: float, t: float) -> float:
    E = 10 * 1e9  # Young's modulus (e.g. 10 GPa)
    load = 40  # load in [kg], distributed evenly over beam
    P = 9.8*load/L  # Distributed load, [N/m]

    Ia = t * H**3 / 12  # area Moment of inertia
    delta_max = (5 * P * L**4) / (384 * E * Ia)

    print(f"{L}m beam loaded with {load}kg (evenly distributed): deflection at center is {delta_max*1e3:.1f}mm")
    
delta_max(1.8, 0.15, 0.012)
delta_max(1.0, 0.04, 0.04)
```
